```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
print(df)
```

```python
print("Shape:")
print(df.shape)

print("Column Names:")
print(df.columns)

print("Data Types:")
print(df.dtypes)

print("Missing Values:")
print(df.isnull().sum())
```

```python
df = df.dropna()
print("After Removing Missing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(5,5))
sns.countplot(x='day', data=df)
plt.title("Count Plot")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
print("Frequency Distribution of Day:")
print(df['day'].value_counts())
print("Frequency Distribution of Sex:")
print(df['sex'].value_counts())
```

```python
print("Percentage Distribution of Day:")
print(df['day'].value_counts()/len(df)*100)
print("Percentage Distribution of Sex:")
print(df['sex'].value_counts()/len(df)*100)
```

```python
print("Most Common Day:")
print(df['day'].mode()[0])
print("Most Common Gender:")
print(df['sex'].mode()[0])
```

```python
plt.pie(df['day'].value_counts(), labels=df['day'].value_counts().index, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
sns.barplot(data=df, x='day', y='total_bill')
plt.title("Bar Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
print(df)
```

```python
print("Shape of Dataset:")
print(df.shape)
```

```python
print("Column Names:")
print(df.columns)
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df = df.dropna()
print("After Removing Missing Values:")
print(df.isnull().sum())
```

```python
sns.countplot(data=df, x='day')
plt.title("Count Plot")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
print("Frequency Distribution:")
print(df['day'].value_counts())
```

```python
print("Percentage Distribution:")
print(df['day'].value_counts(normalize=True)*100)
```

```python
print("Most Common Category:")
print(df['day'].mode())
```

```python
plt.pie(df['day'].value_counts(), labels=df['day'].value_counts().index, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
sns.barplot(data=df, x='day', y='total_bill')
plt.title("Bar Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
sns.countplot(data=df, x='day', hue='sex')
plt.title("Day and Gender Distribution")
plt.show()
```

```python
sns.countplot(data=df, x='day', hue='smoker')
plt.title("Day and Smoker Distribution")
plt.show()
```

```python
print("Mean:")
print(df['total_bill'].mean())

print("Median:")
print(df['total_bill'].median())

print("Mode:")
print(df['total_bill'].mode())

print("Standard Deviation:")
print(df['total_bill'].std())
```

```python
plt.hist(df['total_bill'])
plt.title("Histogram")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(data=df, y='total_bill')
plt.title("Box Plot")
plt.show()
```

```python
sns.violinplot(data=df, y='total_bill')
plt.title("Violin Plot")
plt.show()
```

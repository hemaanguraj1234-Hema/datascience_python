```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("C:/Users/Hema/Downloads/hotel_bookings.csv/hotel_bookings.csv")
data.head()
```

```python
print("Shape of Dataset:")
print(data.shape)
```

```python
print("Column Names:")
print(data.columns)
```

```python
print("Data Types:")
print(data.dtypes)
```

```python
print("Missing Values:")
print(data.isnull().sum())
```

```python
print("Removing Duplicated Values:")
data.duplicated().head()
```

```python
df = data.dropna()
print("After Removing Missing Values:")
print(df.isnull().sum())
```

```python
numeric_cols=data.select_dtypes(include=np.number).columns
print(numeric_cols)
```

```python
for col in numeric_cols:
    plt.figure(figsize=(12,8))
    plt.hist(data[col], bins=5, color='blue', edgecolor='white')
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.show()
```

```python
for col in numeric_cols:
    plt.figure(figsize=(8,5))
    sns.boxplot(x=data[col])
    plt.title(f'Box Plot of {col}')
    plt.show()
```

```python
new = pd.DataFrame({
    "Skewness": data[numeric_cols].skew()
})
print(new.round(3))
```

```python
new = pd.DataFrame({
    "Kurtosis": data[numeric_cols].kurt()
})
print(new.round(3))
```

```python
for col in numeric_cols:
    if data[col].skew() > 0:
        print(col, "- Positively Skewed")
    elif data[col].skew() < 0:
        print(col, "- Negatively Skewed")
    else:
        print(col, "- Normally Distributed")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/Hema/Downloads/flights.csv/flights.csv")
print(df)
```

```python
print("Shape of Dataset:")
print(df.shape)
```

```python
print("Column Names:")
print(df.columns)
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df = df.dropna()
print("After Removing Missing Values:")
print(df.isnull().sum())
```

```python
sns.countplot(data=df, x='month')
plt.title("Count Plot")
plt.xlabel("Month")
plt.ylabel("Count")
plt.show()
```

```python
print("Frequency Distribution:")
print(df['month'].value_counts())
```

```python
print("Percentage Distribution:")
print(df['month'].value_counts(normalize=True)*100)
```

```python
print("Most Common Category:")
print(df['month'].mode())
```

```python
plt.pie(df['month'].value_counts(), autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
freq = df['month'].value_counts()

sns.barplot(x=freq.index, y=freq.values)
plt.title("Bar Chart")
plt.xlabel("Month")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.countplot(data=df, x='month', hue='year', palette='Set2')
plt.title("Month and Year Distribution")
plt.xlabel("Month")
plt.ylabel("Count")
plt.show()
```

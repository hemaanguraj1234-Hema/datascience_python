```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("C:/Users/Hema/OneDrive/Documents/hotel_bookings.csv")
print(data.head())
```

```python
print(data.shape)
```

```python
print(data.columns)
```

```python
print(data.dtypes)
```

```python
print(data.isnull().sum())
```

```python
data = data.dropna()
data['agent'] = data['agent'].fillna(data['agent'].mean())
data['country'] = data['country'].fillna(data['country'].mode()[0])
data['children'] = data['children'].fillna(data['children'].mean())
print(data.isnull().sum())
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='lead_time', y='adr', data=data)
plt.title('Lead Time vs ADR')
plt.xlabel('Lead Time')
plt.ylabel('ADR')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(x='hotel', y='adr', data=data)
plt.title('Hotel Type vs ADR')
plt.xlabel('Hotel Type')
plt.ylabel('ADR')
plt.show()
```

```python
print("Skewness:", data['adr'].skew())
plt.figure(figsize=(8,5))
sns.boxplot(x='hotel', y='adr', data=data)
plt.title('Box Plot with Skewness')
plt.xlabel('Hotel Type')
plt.ylabel('ADR')
plt.show()
```

```python
plt.figure(figsize=(4,6))
sns.barplot(x='hotel', y='adr', data=data)
plt.title('Average ADR by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Average ADR')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(data[['lead_time','adr','adults','children']].corr(), annot=True, cmap='coolwarm')
plt.title("HEAT MAP")
plt.show()
```

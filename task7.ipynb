```python
Q1=numeric_ds.quantile(0.25)
Q3=numeric_ds.quantile(0.75)

IQR=Q3-Q1

lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR

outliers=(
    (numeric_ds < lower_limit) | (numeric_ds > upper_limit)
)

print("Lower Bound:\n", lower_limit)
print("upper Bound:\n", upper_limit)
print("Outliers in each column:\n", numeric_ds[outliers].count())    

```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=sns.load_dataset('flights')
data.head()
```

```python
print(data.head())
print(data.shape)
print(data.columns)
print(data.dtypes)
print(data.isnull().sum())
```

```python
data=data.dropna()
data['month']=data['month'].astype('category')
print(data.isnull().sum())
```

```python
data['month']=data['month'].cat.codes
print(data.head())
```

```python
plt.figure(figsize=(6,6))
sns.scatterplot(x='year',y='passengers',data=data)
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.title('Scatter Plot')
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.violinplot(x='month',y='passengers',data=data)
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.title('Violin Plot')
plt.show()
```

```python
print("Skewness:",data['passengers'].skew())
plt.figure(figsize=(6,6))
sns.boxplot(y='passengers',data=data)
plt.title('Box Plot')
plt.ylabel('Passengers')
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.barplot(x='month',y='passengers',data=data)
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.title('Bar Plot')
plt.show()
```

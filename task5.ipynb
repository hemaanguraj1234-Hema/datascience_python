```python
hemii["Quantity"]=hemii["Quantity"].rolling(window=5).mean()
```

```python
value1=hemii["Quantity"].iloc[0]
value2=hemii["Quantity"].iloc[-1]

if value2 > value1:
    print("Increasing Trend")
elif value2 < value1:
    print("Decreasing Trend")
else:
    print("Stable Trend")
```
